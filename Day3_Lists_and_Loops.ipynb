```python
# 1. THE LIST: Storing multiple tumor sizes (in millimeters) from different patients
tumor_sizes = [12.5, 24.1, 8.9, 32.4, 15.2, 45.0, 7.3]

# 2. THE LOOP: Let's automatically inspect every single tumor size in our list
print("--- Automated Tumor Size Screening ---")

for size in tumor_sizes:
    # If a tumor is larger than 20mm, it's flagged as high risk
    if size > 20.0:
        print(f"Danger! Tumor size {size} mm requires immediate biopsy.")
    else:
        print(f"Safe: Tumor size {size} mm is under monitoring.")

# My try.
patient_ages = [34, 62, 28, 71, 55, 60, 90, 49]

print("--- Clinical Trial Eligibility---")

for age in patient_ages:
  if age > 50:
    print(f"Eligible for Clinical Trials since age is {age}.")
  else:
    print(f"Not Eligible for Clinical Trials since age is {age}.")
```
